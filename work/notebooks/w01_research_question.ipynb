## 2. The question: decision, action, cost of a wrong call
What decision does your work improve? Who acts on it? What does a wrong recommendation cost?

### Research Question
Can we predict which webpages are likely to experience declining search performance so that they can be refreshed before traffic decreases?

**Decision**
The decision is which webpages should be prioritized for content updates.

**Who takes the action?**
The SEO or content team uses the prediction to decide which pages should be refreshed first.

**Action**
Selected pages can be updated by improving content, adding new information, optimizing keywords, or enhancing metadata before their search performance declines.

**Cost of a Wrong Recommendation**
- **False Positive:** A page is updated even though it would have continued performing well — wastes time and resources.
- **False Negative:** A page that actually needs an update is ignored, resulting in reduced search traffic and missed opportunities.

**Why Machine Learning?**
The dataset contains historical search performance data for many webpages. Machine learning can identify patterns in this data and rank pages by likelihood of performance decline, supporting — not replacing — human decisions.

### Quick Look at the Data
The starter dataset contains **30,000 rows and 44 columns**, with one row per page (`content_id` is unique across all rows — no duplicate pages).

Missingness is uneven and worth naming rather than glossing over: `provider_used` is missing in **71.5%** of rows (likely means "no AI tool was used" rather than a data error), and `word_count` / `char_count` are missing in **25.7%** of rows. These will need explicit handling (e.g. a missing-flag category, or imputation) before modeling — they are not just noise to ignore.

These numbers suggest the dataset is large enough to explore patterns in webpage performance, but the missingness pattern is itself a signal worth investigating before building a model.

```python
import pandas as pd

# FIX: use a relative path (works in Colab after upload/mount) instead of a
# hardcoded local Windows path like "C:/Users/Administrator/Downloads/..."
df = pd.read_csv("content_refresh_anonymized.csv")

print("Dataset Shape:", df.shape)
print("Total Records:", len(df))
print("Total Columns:", len(df.columns))
print()
print("Unique Pages:", df["content_id"].nunique())
print()

# Missing values, sorted, with percentage -- more useful than a raw isnull().sum() dump
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(1)
print("Top 10 columns by missing values:")
for col in missing.head(10).index:
    print(f"  {col:<20} {missing[col]:>6} missing  ({missing_pct[col]}%)")
```
